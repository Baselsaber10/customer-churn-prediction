```python
rows, columns = df.shape

missing_values = df.isnull().sum().sum()
duplicate_rows = df.duplicated().sum()
empty_strings = (df == "").sum().sum()
incorrect_dtype = df["TotalCharges"].dtype
```

```python
quality_report = pd.DataFrame({
    "Assessment": [
        "Dataset Shape",
        "Missing Values",
        "Duplicate Rows",
        "Empty Strings",
        "Data Types",
        "Target Variable"
    ],

    "Result": [
        f"{rows} Rows × {columns} Columns",
        missing_values,
        duplicate_rows,
        empty_strings,
        incorrect_dtype,
        "Churn"
    ],

    "Status": [
        "Passed",
        "Passed" if missing_values == 0 else "Failed",
        "Passed" if duplicate_rows == 0 else "Failed",
        "Passed" if empty_strings == 0 else "Failed",
        "Needs Attention" if incorrect_dtype == "object" else "Passed",
        "Passed"
    ],

    "Remarks": [
        "Dataset loaded successfully",
        "No missing values detected",
        "No duplicate records found",
        "No empty strings detected",
        "TotalCharges should be numeric",
        "Binary classification target"
    ]
})

quality_report
```
